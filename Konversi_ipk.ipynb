```python
# konversi_ipk.py
# Program pemroses transkrip nilai akademik

# Bobot penilaian
BOBOT_TUGAS = 0.20
BOBOT_UTS = 0.35
BOBOT_UAS = 0.45

# 1. Terima input satu baris, lalu pisahkan dengan .split(",")
data = input("Masukkan data (nama, NIM, nilai_tugas, nilai_uts, nilai_uas): ")
nama, nim, nilai_tugas, nilai_uts, nilai_uas = data.split(",")

# Hilangkan spasi di awal/akhir tiap bagian
nama = nama.strip()
nim = nim.strip()

# 2. Konversi ketiga nilai ke float
nilai_tugas = float(nilai_tugas)
nilai_uts = float(nilai_uts)
nilai_uas = float(nilai_uas)

# 3. Hitung nilai akhir berdasarkan bobot
nilai_akhir = (nilai_tugas * BOBOT_TUGAS
               + nilai_uts * BOBOT_UTS
               + nilai_uas * BOBOT_UAS)

# 4. Tampilkan tabel dalam satu frame kotak
LEBAR = 44  # lebar isi tabel (di antara dua garis tegak)
garis = "+" + "-" * LEBAR + "+"


def baris(teks):
    """Cetak satu baris tabel dengan batas kiri dan kanan."""
    print("|" + teks.ljust(LEBAR) + "|")


print()
print(garis)
baris(" TRANSKRIP NILAI AKADEMIK".ljust(LEBAR))
print(garis)
baris(f" Nama  : {nama}")
baris(f" NIM   : {nim}")
print(garis)
baris(f" {'Komponen':<12}{'Nilai':>10}{'Bobot':>12}")
print(garis)
baris(f" {'Tugas':<12}{nilai_tugas:>10.1f}{BOBOT_TUGAS:>12.0%}")
baris(f" {'UTS':<12}{nilai_uts:>10.1f}{BOBOT_UTS:>12.0%}")
baris(f" {'UAS':<12}{nilai_uas:>10.1f}{BOBOT_UAS:>12.0%}")
print(garis)
baris(f" {'NILAI AKHIR':<12}{nilai_akhir:>10.1f}")
print(garis)
```
